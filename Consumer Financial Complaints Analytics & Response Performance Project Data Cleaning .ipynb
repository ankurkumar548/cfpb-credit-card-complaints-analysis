```python
df.columns=(
    df.columns
    .str.strip()
    .str.lower()
    .str.replace("?","", regex=False)
    .str.replace(" ", "_",regex=False)
    .str.replace("-", "_", regex=False)
)
print(df.columns)
```

```python
df=df.drop_duplicates(subset=["complaint_id"], keep="first")
```

```python
df["date_received"]=pd.to_datetime(df["date_received"])
df["date_sent_to_company"]=pd.to_datetime(df["date_sent_to_company"])
```

```python
date_columns=[
    "date_received",
    "date_sent_to_company"
]
```

```python
df[date_columns].isna().sum()
```

```python
df=df[
    (df["date_received"]>="2024-01-01")&
    (df["date_received"]<="2025-01-01")
].copy()
```

```python
categorical_columns=[
"sub_issue",                        
"consumer_complaint_narrative",   
"company_public_response","state","tags"    
]

df[categorical_columns]=(df[categorical_columns].fillna("Not Provided"))
```

```python
df["zip_code"] = (
    df["zip_code"]
    .astype(str)
    .str.strip()
    .str.replace(".0", "", regex=False)
)

df.loc[
    ~df["zip_code"].str.match(r"^\d{3,5}$", na=False),
    "zip_code"
] = "Not Provided"
```

```python
df["timely_response"] = (
    df["timely_response"]
    .str.strip()
    .str.title()
)

df.loc[
    ~df["timely_response"].isin(["Yes", "No"]),
    "timely_response"
] = "Unknown"
```

```python
df["received_year"] = df["date_received"].dt.year
df["received_month"] = df["date_received"].dt.month
df["month_name"] = df["date_received"].dt.month_name()
df["received_quarter"] = "Q" + df["date_received"].dt.quarter.astype(str)
df["day_name"] = df["date_received"].dt.day_name()

df["days_to_send"] = (
    df["date_sent_to_company"] - df["date_received"]
).dt.total_seconds() / 86400
```

```python
print(df["days_to_send"].describe())

invalid = df[df["days_to_send"] < 0]
print("Invalid date sequence:", len(invalid))
```

```python
df.loc[df["days_to_send"] < 0, "days_to_send"] = pd.NA
```

```python
print("Final rows:", len(df))
print("Duplicate IDs:", df["complaint_id"].duplicated().sum())
print("Missing values:\n", df.isnull().sum())
print("Date range:", df["date_received"].min(), df["date_received"].max())
print(df["timely_response"].value_counts())
```
